```python
data_with_missing = {
    "cars": ["BMW", "Volvo", None, "Tesla", "Ferrari", None, "Tayota"],
    "counts": [3, 7, None, 8, 1, 9, None ]
}

df_missing = pd.DataFrame(data_with_missing)
print(df_missing)
```

```python
# df_missing.isnull() # boolean qaytaradi, jadvaldagi bor ma'lumotni false, yo'q ma'lumotni true qaytaradi
df_missing.isnull().sum() # har bir columnda nechta data missingligini aniqlab beradi
```

```python
df_missing.dropna() # bo'sh qatorlarni jadvaldan o'chirib beradi
```

```python
df_missing.dropna(how="all") # barcha qiymatlari bo'sh bo'lgan qatorni o'chiradi
```

```python
df_missing.dropna(axis=1) # ma'lumoti bo'sh bo'lgan barcha columnlarni o'chiradi
```

```python
df_missing.dropna(subset=['counts']) # berilgan columnigina tekshiradi va unda biror ma'lumot kam bo'lsa shuni tashlab yuboradi
```

```python
df_missing.fillna(0) # bo'sh qatorlarga biz bergan argumentni yozadi
```

```python
df_missing.fillna({"cars": "unknown", "counts": 0}) # kerakli columnlarga alohida missing ma'lumotlarni to'ldirish
```
